## Modelo de regresión

<img align="left" src="./modelo_lr.png" style="width:380px; padding:10px;"> Como se describió en la sesión técnica, la función del modelo para la regresión lineal (que es una función que mapea valores de `x` a valores de `y`) se representa de la siguiente manera:

$$ f_{w,b}(x^{(i)}) = wx^{(i)} + b \tag{1}$$

La fórmula anterior permite representar líneas rectas. Diferentes valores de $w$ y $b$ producen diferentes líneas rectas en el gráfico. <br>

Para comprenderlo mejor, probaremos distintos valores en los siguientes bloques de código. Comenzaremos con $w = 100$ y $b = 100$.

**Nota: Puedes volver a esta celda para ajustar los parámetros $w$ y $b$ del modelo.**

```python
w = 100
b = 100
print(f"w: {w}")
print(f"b: {b}")
```

Ahora calcularemos el valor de $f_{w,b}(x^{(i)})$ para nuestros dos puntos de datos. Puedes escribirlo explícitamente para cada punto de la siguiente manera:

para $x^{(0)}$, `f_wb = w * x[0] + b`

para $x^{(1)}$, `f_wb = w * x[1] + b`

Cuando hay una gran cantidad de puntos de datos, este procedimiento puede volverse complicado y repetitivo. Por eso, podemos calcular la salida de la función dentro de un bucle `for`, como se muestra en la función `compute_model_output` de la siguiente celda.

> **Nota**: La descripción del argumento `(ndarray (m,))` indica un arreglo n-dimensional de NumPy con forma `(m,)`. `(scalar)` describe un argumento sin dimensiones, es decir, un valor único.
> **Nota**: `np.zeros(n)` devuelve un arreglo unidimensional de NumPy con $n$ elementos.

```python
def compute_model_output(x, w, b):
    """
    Computes the prediction of a linear model
    Args:
      x (ndarray (m,)): Data, m examples 
      w,b (scalar)    : model parameters  
    Returns
      f_wb (ndarray (m,)): model prediction
    """
    m = x.shape[0]
    f_wb = np.zeros(m)
    for i in range(m):
        f_wb[i] = w * x[i] + b
        
    return f_wb
```

Ahora invocaremos a la función `compute_model_output` y graficaremos la salida..

```python
tmp_f_wb = compute_model_output(x_train, w, b,)

# Plot our model prediction
plt.plot(x_train, tmp_f_wb, c='b',label='Our Prediction')

# Plot the data points
plt.scatter(x_train, y_train, marker='x', c='r',label='Actual Values')

# Set the title
plt.title("Housing Prices")
# Set the y-axis label
plt.ylabel('Price (in 1000s of dollars)')
# Set the x-axis label
plt.xlabel('Size (1000 sqft)')
plt.legend()
plt.show()
```

Como podemos observar, estableciendo a $w = 100$ y $b = 100$  *no resulta* en unalinea que se ajusta a nuestros datos.

Ahora que tenemos un modelo, podemos utilizarlo para realizar nuestra predicción inicial.
