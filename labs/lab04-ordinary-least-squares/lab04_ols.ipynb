# Lab 03 – Mínimos Cuadrados Ordinarios (OLS) para regresión lineal univariada

Modelo: $y=\beta_0+\beta_1x$

OLS elige $\beta_0,\beta_1$ minimizando la suma de errores cuadrados $\sum (y_i-\hat y_i)^2$. La solución cerrada es:

$$\beta_1=\frac{\sum_{i=1}^{n}(x_i-\bar{x})(y_i-\bar{y})}{\sum_{i=1}^{n}(x_i-\bar{x})^2}\qquad \beta_0=\bar{y}-\beta_1\bar{x}$$

En este notebook implementarás todo **paso a paso y sin vectorizar** (con ciclos `for`). En cada reto escribe tu código en la celda; si te atoras, abre el desplegable con la solución.

## Reto 3. Medias
Calcula $\bar{x}$ y $\bar{y}$ **sin** `np.mean`: usa un ciclo `for` para sumar y divide entre $n$.

````python
# TU CÓDIGO AQUÍ
n = None
x_mean = None
y_mean = None
````

<details>
<summary>💡 Ver solución</summary>

```python
n = len(X)
suma_x = 0.0
suma_y = 0.0
for i in range(n):
    suma_x += X[i, 0]
    suma_y += Y[i, 0]
x_mean = suma_x / n
y_mean = suma_y / n
print(x_mean, y_mean)
```

Acumulamos la suma elemento a elemento y dividimos entre $n$. `X[i, 0]` extrae el escalar de la fila `i`. Debe coincidir con `X.mean()` (≈1) y `Y.mean()` (≈7.5).

</details>

## Reto 4. Numerador de $\beta_1$
Calcula $S_{xy}=\sum_{i=1}^{n}(x_i-\bar{x})(y_i-\bar{y})$ con un ciclo `for`.

````python
# TU CÓDIGO AQUÍ
Sxy = None
````

<details>
<summary>💡 Ver solución</summary>

```python
Sxy = 0.0
for i in range(n):
    Sxy += (X[i, 0] - x_mean) * (Y[i, 0] - y_mean)
print(Sxy)
```

Es la covarianza (sin dividir entre $n$): mide cómo varían juntas $x$ y $y$ respecto a sus medias. Es positiva porque ambas crecen juntas.

</details>

## Reto 5. Denominador de $\beta_1$
Calcula $S_{xx}=\sum_{i=1}^{n}(x_i-\bar{x})^2$.

````python
# TU CÓDIGO AQUÍ
Sxx = None
````

<details>
<summary>💡 Ver solución</summary>

```python
Sxx = 0.0
for i in range(n):
    Sxx += (X[i, 0] - x_mean) ** 2
print(Sxx)
```

Es la variación total de $x$ alrededor de su media (varianza sin dividir entre $n$). Normaliza la covarianza para obtener la pendiente.

</details>

## Reto 6. Calcular $\beta_1$ y $\beta_0$
Usa $S_{xy}$, $S_{xx}$ y las medias. Recuerda: primero $\beta_1$, luego $\beta_0=\bar y-\beta_1\bar x$.

````python
# TU CÓDIGO AQUÍ
beta1 = None
beta0 = None
print(f'beta0 = {beta0}, beta1 = {beta1}')
````

<details>
<summary>💡 Ver solución</summary>

```python
beta1 = Sxy / Sxx
beta0 = y_mean - beta1 * x_mean
print(f'beta0 = {beta0:.4f}, beta1 = {beta1:.4f}')
```

$\beta_1\approx3$ (pendiente real) y $\beta_0\approx4.5$ (4 + media del ruido 0.5). **Ojo:** la fracción del enunciado es $\beta_1$, y es ésta la que multiplica a $\bar x$: $\beta_0=\bar y-\beta_1\bar x$.

</details>
