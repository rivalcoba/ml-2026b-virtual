## Cálculo del costo
El término «costo» puede resultar confuso porque los datos se refieren al costo de las viviendas. Aquí, el costo mide qué tan bien predice nuestro modelo el precio de una casa. Para los datos de viviendas, usaremos el término «precio».

La ecuación del costo para una variable es:
  $$J(w,b) = \frac{1}{2m} \sum_{i=1}^{m} \left(f_{w,b}(x^{(i)}) - y^{(i)}\right)^2 \tag{1}$$
 
donde
  $$f_{w,b}(x^{(i)}) = wx^{(i)} + b \tag{2}$$
  
- $f_{w,b}(x^{(i)})$ es la predicción para el ejemplo $i$, usando los parámetros $w$ y $b.
- $\left(f_{w,b}(x^{(i)}) - y^{(i)}\right)^2$ es el cuadrado de la diferencia entre el valor objetivo y la predicción.
- Estas diferencias se suman para los $m$ ejemplos y se dividen entre $2m$ para obtener el costo $J(w,b)$.

> En la notación matemática, los ejemplos se numeran de $1$ a $m$. En Python, los índices de los arreglos van de $0$ a $m-1$.

El siguiente código calcula el costo recorriendo cada ejemplo. En cada iteración:

- Se calcula `f_wb`, la predicción.
- Se calcula la diferencia entre el valor objetivo y la predicción, y se eleva al cuadrado.
- Este valor se suma al costo total.

```python
def compute_cost(x, y, w, b): 
    """
    Computes the cost function for linear regression.
    
    Args:
      x (ndarray (m,)): Data, m examples 
      y (ndarray (m,)): target values
      w,b (scalar)    : model parameters  
    
    Returns
        total_cost (float): The cost of using w,b as the parameters for linear regression
               to fit the data points in x and y
    """
    # number of training examples
    m = x.shape[0] 
    
    cost_sum = 0 
    for i in range(m): 
        f_wb = w * x[i] + b   
        cost = (f_wb - y[i]) ** 2  
        cost_sum = cost_sum + cost  
    total_cost = (1 / (2 * m)) * cost_sum  

    return total_cost
```

## Intuición de la función de costo

El objetivo es encontrar un modelo $f_{w,b}(x) = wx + b$, con parámetros $w$ y $b$, que prediga con precisión el valor de las viviendas a partir de una entrada $x$. El costo mide qué tan preciso es el modelo con los datos de entrenamiento.

La ecuación del costo (1) muestra que, si se pueden elegir $w$ y $b$ de modo que las predicciones $f_{w,b}(x)$ coincidan con los valores objetivo $y$, el término $\left(f_{w,b}(x^{(i)}) - y^{(i)}\right)^2$ será cero y el costo se minimizará. ¡En este sencillo ejemplo de dos puntos, esto es posible!

En el laboratorio anterior determinaste que $b=100$ daba una solución óptima, así que fijaremos $b$ en 100 y nos centraremos en $w$.

### Superficie de costo convexa

El hecho de que la función de costo eleve la pérdida al cuadrado garantiza que la «superficie de error» sea convexa, como un tazón de sopa. Siempre tendrá un mínimo al que se puede llegar siguiendo el gradiente en todas las dimensiones. En la gráfica anterior, como las dimensiones $w$ y $b$ tienen escalas distintas, esto no es fácil de reconocer. La siguiente gráfica, en la que $w$ y $b$ son simétricas, se mostró en clase:
